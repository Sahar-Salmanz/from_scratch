# Develpoing Linear Regression Algorithm from Scratch

Linear regression is about finding the function that minimizes the error (the difference between the real target value and the predicted value)

```python
# import pandas for loading a csv file
import pandas as pd
# import matplotlib for visualization
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('house_prices.csv')
```

```python
data.head(5)
```

```python
plt.scatter(data['Size_sqft'], data['Price_USD'])
plt.xlabel('Size (sqft)')
plt.ylabel('Price (USD)')
plt.title('House Prices vs Size')
```

```python
def loss_function(m, b, data):
    total_loss = 0
    for i in range(len(data)):
        x = data['Size_sqft'][i]
        y = data['Price_USD'][i]
        total_loss += (y - (m * x + b)) ** 2
    return total_loss / len(data)
```

```python
def gradient_descent(m_old, b_old, data, L):
    m_gradient = 0
    b_gradient = 0
    n = len(data)

    for i in range(n):
        x = data.iloc[i].Size_sqft
        y = data.iloc[i].Price_USD

        m_gradient += -(2/n) * x * (y - (m_old * x + b_old))
        b_gradient += -(2/n) * (y - (m_old * x + b_old))

    m = m_old - (L * m_gradient)
    b = b_old - (L * b_gradient)

    return m, b
```

```python
m = 0
b = 0
L = 0.0001
iterations = 1000

for i in range(iterations):
    m, b = gradient_descent(m, b, data, L)

print(m, b)
```
